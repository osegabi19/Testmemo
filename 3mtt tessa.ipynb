**Conditional Statement**

and, or and non

```python
age = 10

if age >= 18:
  print("You are eligible to vote")
else:
  print("You are not eligible to vote")
```

```python
grade = 80
comment = "you got an "
comments = "you got a "

if grade >= 90:
  print(f"{comment} A")
  print("Excellent")
elif grade >= 80:
  print(f"{comments} B")
  print("Good")
elif grade >= 70:
  print(f"{comments} C")
  print("Average")
elif grade >= 60:
  print(f"{comments} D")
  print("Poor")
else:
  print("Repeat the exam")
```

```python
grade = int(input("What is your grade in your English Test? "))
comment = "you got a "

if grade >= 90:
  print(f"{comment} A")
  print("Excellent")
elif grade >= 80:
  print(f"{comment} B")
  print("Good")
elif grade >= 70:
  print(f"{comment} C")
  print("Average")
elif grade >= 60:
  print(f"{comment} D")
  print("Poor")
else:
  print(f"Repeat the exam, your {grade} is too low")
```
